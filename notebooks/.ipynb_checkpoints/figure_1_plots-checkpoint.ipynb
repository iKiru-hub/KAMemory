# Figure 1 plot generator

This notebook generates every draft panel sketched in `plot_ideas/figure_1_idea.jpeg` and saves the outputs in `media/generated/figure_1/`.

The simulation cells use a lightweight NumPy implementation of the same BTSP-style update used in the project: CA3 activity provides the eligibility trace, the EC-to-CA1 content code provides the instructive signal, and CA1-to-EC readout is held fixed.

```python
from pathlib import Path
import json
import math
import os

os.environ.setdefault("MPLCONFIGDIR", "/private/tmp/kamemory_mplconfig")
Path(os.environ["MPLCONFIGDIR"]).mkdir(parents=True, exist_ok=True)

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib import gridspec
from matplotlib.patches import Circle, Rectangle, FancyArrowPatch
from PIL import Image


def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start)
    for path in [start, *start.parents]:
        if (path / "README.md").exists() and (path / "src").exists():
            return path
    raise RuntimeError("Could not find KAMemory project root")


ROOT = find_project_root()
OUTDIR = ROOT / "media" / "generated" / "figure_1"
OUTDIR.mkdir(parents=True, exist_ok=True)
IDEA_IMAGE = ROOT / "plot_ideas" / "figure_1_idea.jpeg"

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 10,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "figure.dpi": 120,
})

COLORS = {
    "ca3": "#202020",
    "ca1": "#202020",
    "ec": "#202020",
    "btsp": "#d83b36",
    "fixed": "#3a3a3a",
    "pretrained": "#2fa36b",
    "aligned": "#1f77b4",
    "random": "#b34a4a",
    "green": "#2fa36b",
}


def panel_label(ax, label):
    ax.text(-0.08, 1.06, label, transform=ax.transAxes, fontsize=18,
            fontweight="bold", va="bottom", ha="left")


def save_panel(fig, name, outdir=OUTDIR):
    path = outdir / name
    fig.savefig(path, dpi=220, bbox_inches="tight", pad_inches=0.15, facecolor="white")
    plt.close(fig)
    print(f"saved {path.relative_to(ROOT)}")
    return path


def sparse_stimuli(rng, n, k, dim):
    samples = np.zeros((n, dim), dtype=float)
    for i in range(n):
        samples[i, rng.choice(dim, size=k, replace=False)] = 1.0
    return samples


def sparsemoid(z, k, beta=60.0):
    z = np.asarray(z, dtype=float)
    original_shape = z.shape
    if z.ndim == 1:
        z = z[None, :]
    k = int(max(1, min(k, z.shape[1] - 1)))
    sorted_z = np.sort(z, axis=1)[:, ::-1]
    theta = sorted_z[:, k - 1:k + 1].mean(axis=1, keepdims=True)
    out = 1.0 / (1.0 + np.exp(-np.clip(beta * (z - theta), -50, 50)))
    return out.reshape(original_shape)


def cosine(a, b):
    a = np.asarray(a).ravel()
    b = np.asarray(b).ravel()
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-9))


def circular_distance(x1, x2, n):
    return np.minimum(np.abs(x1 - x2), n - np.abs(x1 - x2))


def place_field_activity(n_x, sigma, xi):
    x = np.arange(n_x)
    return np.exp(-(circular_distance(x, xi, n_x) ** 2) / (2 * sigma ** 2))


def prepare_lightweight_model(seed=0, dim_ei=50, dim_ca1=80, dim_ca3=80,
                              k_ei=5, k_ca1=12, k_ca3=12, train_n=700):
    """Build a small NumPy version of the KAMemory equations for plot generation."""
    rng = np.random.default_rng(seed)
    training = sparse_stimuli(rng, train_n, k_ei, dim_ei)

    w_ei_ca1 = rng.normal(0, 1 / np.sqrt(dim_ei), size=(dim_ca1, dim_ei))
    z_ca1 = sparsemoid(training @ w_ei_ca1.T, k_ca1, beta=60)

    ridge = 1e-2
    decoder_fit = np.linalg.solve(
        z_ca1.T @ z_ca1 + ridge * np.eye(dim_ca1),
        z_ca1.T @ training,
    )
    w_ca1_eo = decoder_fit.T
    w_ei_ca3 = rng.normal(0, 1 / np.sqrt(dim_ei), size=(dim_ca3, dim_ei))

    return {
        "rng": rng,
        "dim_ei": dim_ei,
        "dim_ca1": dim_ca1,
        "dim_ca3": dim_ca3,
        "k_ei": k_ei,
        "k_ca1": k_ca1,
        "k_ca3": k_ca3,
        "w_ei_ca1": w_ei_ca1,
        "w_ca1_eo": w_ca1_eo,
        "w_ei_ca3": w_ei_ca3,
    }


def make_row_shuffled_encoder(rng, w_ei_ca1, fraction):
    w_is = w_ei_ca1.copy()
    n_rows = w_is.shape[0]
    n_shuffle = int(round(fraction * n_rows))
    if n_shuffle > 1:
        rows = rng.choice(n_rows, size=n_shuffle, replace=False)
        permuted = rows.copy()
        rng.shuffle(permuted)
        w_is[rows] = w_is[permuted]
    return w_is


def run_memory(seed=1, n_patterns=30, alpha=0.35, shuffle_fraction=0.0):
    model = prepare_lightweight_model(seed=seed)
    rng = model["rng"]
    samples = sparse_stimuli(rng, n_patterns, model["k_ei"], model["dim_ei"])
    w_ca3_ca1 = np.zeros((model["dim_ca1"], model["dim_ca3"]))
    w_is = make_row_shuffled_encoder(rng, model["w_ei_ca1"], shuffle_fraction)

    accuracy = np.full((n_patterns, n_patterns), np.nan)
    final_outputs = []
    alignments = []
    ca1_snapshots = []

    for i, x in enumerate(samples):
        x_ca3 = sparsemoid(model["w_ei_ca3"] @ x, model["k_ca3"], beta=120)
        is_true = sparsemoid(model["w_ei_ca1"] @ x, model["k_ca1"], beta=60)
        is_used = sparsemoid(w_is @ x, model["k_ca1"], beta=60)
        alignments.append(cosine(is_true, is_used))

        w_ca3_ca1 = (
            (1 - alpha * is_used[:, None]) * w_ca3_ca1
            + alpha * (is_used[:, None] @ x_ca3[None, :])
        )

        for j, x_j in enumerate(samples[:i + 1]):
            x3_j = sparsemoid(model["w_ei_ca3"] @ x_j, model["k_ca3"], beta=120)
            ca1_j = sparsemoid(w_ca3_ca1 @ x3_j, model["k_ca1"], beta=120)
            y_j = sparsemoid(model["w_ca1_eo"] @ ca1_j, model["k_ei"], beta=60)
            accuracy[i, j] = cosine(x_j, y_j)

        if i in {4, 14, n_patterns - 1}:
            ca1_snapshots.append((i, w_ca3_ca1.copy()))

    for x_j in samples:
        x3_j = sparsemoid(model["w_ei_ca3"] @ x_j, model["k_ca3"], beta=120)
        ca1_j = sparsemoid(w_ca3_ca1 @ x3_j, model["k_ca1"], beta=120)
        final_outputs.append(sparsemoid(model["w_ca1_eo"] @ ca1_j, model["k_ei"], beta=60))

    return {
        "model": model,
        "samples": samples,
        "accuracy": accuracy,
        "final_outputs": np.array(final_outputs),
        "alignment": float(np.mean(alignments)),
        "ca1_snapshots": ca1_snapshots,
        "w_ca3_ca1": w_ca3_ca1,
    }


def retention_curve(accuracy):
    n = accuracy.shape[0]
    means = []
    sems = []
    for age in range(n):
        vals = []
        for step in range(n):
            j = step - age
            if j >= 0 and not np.isnan(accuracy[step, j]):
                vals.append(accuracy[step, j])
        vals = np.array(vals)
        means.append(np.nanmean(vals))
        sems.append(np.nanstd(vals) / np.sqrt(max(1, len(vals))))
    return np.arange(n), np.array(means), np.array(sems)
```

## Run memory simulations

The aligned condition uses the CA1 content code as the instructive signal. The shuffled condition permutes CA1 rows in the instructive-signal encoder, breaking the alignment between the IS and the fixed CA1 decoder.

```python
aligned = run_memory(seed=12, n_patterns=30, alpha=0.35, shuffle_fraction=0.0)
shuffled = run_memory(seed=12, n_patterns=30, alpha=0.35, shuffle_fraction=1.0)

np.savez(
    OUTDIR / "figure_1_simulation.npz",
    aligned_accuracy=aligned["accuracy"],
    shuffled_accuracy=shuffled["accuracy"],
    aligned_samples=aligned["samples"],
    aligned_outputs=aligned["final_outputs"],
    shuffled_outputs=shuffled["final_outputs"],
)
print("aligned final mean decodability:", np.nanmean(aligned["accuracy"][-1]))
print("shuffled final mean decodability:", np.nanmean(shuffled["accuracy"][-1]))
```

## A. BTSP-induced weight change

```python
def plot_panel_a():
    fig = plt.figure(figsize=(7.6, 3.2))
    gs = gridspec.GridSpec(1, 2, width_ratios=[1.05, 1.2], wspace=0.35)
    ax = fig.add_subplot(gs[0, 0])
    ax.set_axis_off()
    panel_label(ax, "A")

    positions = {"ca3": (0.12, 0.58), "ca1": (0.55, 0.58), "ec": (0.90, 0.58)}
    for key, (x, y) in positions.items():
        circ = Circle((x, y), 0.09, fill=False, lw=1.8, color="black")
        ax.add_patch(circ)
    ax.add_patch(FancyArrowPatch((0.21, 0.58), (0.46, 0.58), arrowstyle="-|>", mutation_scale=14, lw=1.6, color="black"))
    ax.add_patch(FancyArrowPatch((0.82, 0.58), (0.64, 0.58), arrowstyle="-|>", mutation_scale=14, lw=1.6, color="black"))
    ax.text(0.31, 0.65, r"$W^{CA1 \leftarrow CA3}_{ij}$", ha="center", fontsize=12)
    ax.text(0.12, 0.38, "CA3\nneuron j", ha="center")
    ax.text(0.55, 0.38, "CA1\nneuron i", ha="center")
    ax.text(0.90, 0.35, "Entorhinal\ncortex", ha="center")
    ax.text(0.70, 0.64, "IS", ha="center", fontsize=11)
    ax.plot([0.03, 0.47], [0.20, 0.20], color="black", lw=1)
    ax.text(0.03, 0.24, "CA1 neuron receives IS", ha="left")
    ax.plot([0.03, 0.47], [0.08, 0.08], color="black", lw=1)
    ax.text(0.03, 0.12, "No IS in CA1 neuron", ha="left")
    ax.add_patch(FancyArrowPatch((0.50, 0.08), (0.70, 0.08), arrowstyle="-|>", mutation_scale=14, lw=1.2, color="black"))
    ax.text(0.73, 0.08, "No BTSP", va="center")
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)

    ax2 = fig.add_subplot(gs[0, 1])
    t = np.linspace(-5, 5, 180)
    v = np.linspace(0, 14, 150)
    T, V = np.meshgrid(t, v)
    kernel = 14 * np.exp(-(T / 1.5) ** 2) * (1 / (1 + np.exp(-(V - 5))))
    kernel -= 8 * np.exp(-((np.abs(T) - 3.2) / 0.8) ** 2) * (1 / (1 + np.exp(-(V - 7))))
    im = ax2.imshow(kernel, extent=[t.min(), t.max(), v.min(), v.max()],
                    origin="lower", aspect="auto", cmap="RdBu_r", vmin=-14, vmax=14)
    ax2.contour(T, V, kernel, colors="0.7", levels=[-5, 0, 5], linewidths=1)
    ax2.axvline(0, color="0.2", lw=1, ls=":")
    ax2.set_xlabel("Time from plateau (s)")
    ax2.set_ylabel("Initial Vm ramp (mV)")
    ax2.set_title("BTSP window schematic", fontsize=11)
    cbar = fig.colorbar(im, ax=ax2, shrink=0.78)
    cbar.set_label(r"$\Delta W$")
    return fig


save_panel(plot_panel_a(), "panel_A_btsp_weight_change.png")
```

## B. Kernel fixed points

```python
def plot_panel_b():
    fig, axes = plt.subplots(1, 2, figsize=(7.0, 2.7), sharey=True)
    panel_label(axes[0], "B")
    w = np.linspace(0, 1, 200)
    alpha = 0.25
    maps = {
        "inactive CA3": (1 - alpha) * w,
        "active CA3": (1 - alpha) * w + alpha,
    }

    for ax, title in zip(axes, ["fixed point intuition", "BTSP update map"]):
        ax.plot(w, w, color="0.75", lw=1, ls="--")
        ax.plot(w, maps["inactive CA3"], color=COLORS["random"], lw=2, label="CA3 inactive")
        ax.plot(w, maps["active CA3"], color=COLORS["aligned"], lw=2, label="CA3 active")
        ax.scatter([0, 1], [0, 1], s=35, color=["#b34a4a", "#1f77b4"], zorder=3)
        ax.annotate("goes to 0", xy=(0.18, maps["inactive CA3"][35]), xytext=(0.42, 0.18),
                    arrowprops=dict(arrowstyle="->", lw=1), fontsize=9)
        ax.annotate("goes to 1", xy=(0.78, maps["active CA3"][155]), xytext=(0.46, 0.82),
                    arrowprops=dict(arrowstyle="->", lw=1), fontsize=9)
        ax.set_xlabel(r"$W_t$")
        ax.set_title(title, fontsize=11)
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
    axes[0].set_ylabel(r"$W_{t+1}$")
    axes[1].legend(frameon=False, loc="lower right", fontsize=8)
    return fig


save_panel(plot_panel_b(), "panel_B_kernel_fixed_points.png")
```

## D. Architecture

```python
def draw_node(ax, xy, label):
    circ = Circle(xy, 0.12, fill=False, lw=2.2, color="black")
    ax.add_patch(circ)
    ax.text(*xy, label, ha="center", va="center", fontsize=12)


def draw_arrow(ax, start, end, color="black", lw=2, style="-|>", ls="-", text=None, text_xy=None):
    ax.add_patch(FancyArrowPatch(start, end, arrowstyle=style, mutation_scale=15,
                                 lw=lw, color=color, linestyle=ls))
    if text:
        ax.text(*text_xy, text, fontsize=9, ha="center", va="center", color=color)


def plot_panel_d():
    fig, ax = plt.subplots(figsize=(4.8, 3.5))
    ax.set_axis_off()
    panel_label(ax, "D")
    ax.text(0.50, 0.95, "Architecture", ha="center", va="top", fontsize=14, fontweight="bold")

    pos = {"CA3": (0.28, 0.62), "CA1": (0.63, 0.62), "ECin": (0.28, 0.24), "ECout": (0.63, 0.24)}
    for label, xy in pos.items():
        draw_node(ax, xy, label.replace("ECin", "EC(in)").replace("ECout", "EC(out)"))
    draw_arrow(ax, (0.40, 0.62), (0.51, 0.62), color=COLORS["btsp"], lw=2.2)
    draw_arrow(ax, (0.28, 0.36), (0.28, 0.50), color=COLORS["fixed"], lw=1.7, ls=":")
    draw_arrow(ax, (0.38, 0.30), (0.53, 0.52), color=COLORS["pretrained"], lw=1.7, ls=":")
    draw_arrow(ax, (0.63, 0.50), (0.63, 0.36), color=COLORS["pretrained"], lw=1.9, ls=":")
    ax.add_patch(Rectangle((0.42, 0.38), 0.13, 0.08, fill=False, lw=1.5))
    ax.text(0.485, 0.42, "IS", ha="center", va="center", fontsize=11)

    ax.plot([0.82, 0.96], [0.78, 0.78], color=COLORS["fixed"], lw=1.8, ls=":")
    ax.text(0.98, 0.78, "fixed random", va="center", fontsize=9)
    ax.plot([0.82, 0.96], [0.66, 0.66], color=COLORS["pretrained"], lw=1.8, ls=":")
    ax.text(0.98, 0.66, "fixed pre-trained", va="center", fontsize=9)
    ax.plot([0.82, 0.96], [0.54, 0.54], color=COLORS["btsp"], lw=2.2)
    ax.text(0.98, 0.54, "BTSP", va="center", fontsize=9)
    ax.set_xlim(0, 1.15)
    ax.set_ylim(0, 1)
    return fig


save_panel(plot_panel_d(), "panel_D_architecture.png")
```

## E. Receptive field formation

```python
def plot_panel_e():
    fig, axes = plt.subplots(3, 1, figsize=(4.6, 3.4), sharex=True)
    panel_label(axes[0], "E")
    x = np.linspace(0, 1, 200)
    centers = [0.35, 0.50, 0.65]
    colors = ["#c7dcef", "#75a9d6", "#1f77b4"]
    for ax, c, color, lap in zip(axes, centers, colors, [1, 6, 18]):
        field = np.exp(-((x - c) ** 2) / (2 * 0.055 ** 2))
        ax.fill_between(x, 0, field, color=color, alpha=0.85)
        ax.plot(x, field, color="black", lw=0.7)
        ax.axvline(0.5, color=COLORS["btsp"], lw=1, ls="--")
        ax.text(0.02, 0.72, f"after lap {lap}", transform=ax.transAxes, fontsize=9)
        ax.set_yticks([])
        ax.set_ylim(0, 1.08)
    axes[-1].set_xlabel("position on track")
    fig.suptitle("diagram of receptive-field formation", fontsize=11, y=0.98)
    return fig


save_panel(plot_panel_e(), "panel_E_receptive_field_formation.png")
```

## H. Example decoded patterns

```python
def plot_panel_h(record):
    n_show = 6
    fig, axes = plt.subplots(3, 1, figsize=(6.5, 3.2), sharex=True)
    panel_label(axes[0], "H")
    matrices = [
        (record["samples"][:n_show], "ECin patterns"),
        (record["final_outputs"][:n_show], "ECout recall"),
        (np.abs(record["samples"][:n_show] - record["final_outputs"][:n_show]), "absolute error"),
    ]
    for ax, (mat, label) in zip(axes, matrices):
        cmap = "gray_r" if label != "absolute error" else "magma"
        ax.imshow(mat, aspect="auto", cmap=cmap, vmin=0, vmax=1)
        ax.set_ylabel(label, rotation=0, ha="right", va="center", labelpad=55)
        ax.set_yticks(range(n_show))
        ax.set_yticklabels(range(1, n_show + 1), fontsize=8)
        ax.set_xticks([])
    axes[-1].set_xlabel("EC unit")
    return fig


save_panel(plot_panel_h(aligned), "panel_H_decoded_pattern_examples.png")
```

## I. Quantifying decodability

```python
def plot_panel_i(aligned, shuffled):
    fig = plt.figure(figsize=(7.8, 3.6))
    gs = gridspec.GridSpec(1, 2, width_ratios=[1.0, 1.15], wspace=0.35)
    ax0 = fig.add_subplot(gs[0, 0])
    panel_label(ax0, "I")
    matrix = np.ma.masked_invalid(aligned["accuracy"])
    im = ax0.imshow(matrix, origin="lower", cmap="viridis", vmin=0, vmax=1, aspect="auto")
    ax0.set_xlabel("memory index")
    ax0.set_ylabel("last stored pattern")
    ax0.set_title("recall quality matrix", fontsize=11)
    cbar = fig.colorbar(im, ax=ax0, shrink=0.8)
    cbar.set_label("cosine similarity")

    ax1 = fig.add_subplot(gs[0, 1])
    for rec, label, color in [
        (aligned, "content-aware IS", COLORS["aligned"]),
        (shuffled, "shuffled IS", COLORS["random"]),
    ]:
        age, mean, sem = retention_curve(rec["accuracy"])
        ax1.plot(age, mean, color=color, lw=2, label=label)
        ax1.fill_between(age, mean - sem, mean + sem, color=color, alpha=0.18, lw=0)
    ax1.set_xlabel("# patterns stored afterwards")
    ax1.set_ylabel("decodability")
    ax1.set_ylim(0, 1.02)
    ax1.set_title("quality decay", fontsize=11)
    ax1.legend(frameon=False, fontsize=9)
    return fig


save_panel(plot_panel_i(aligned, shuffled), "panel_I_decodability_quantification.png")
```

## Assemble draft Figure 1

```python
def assemble_composite(panel_names, out_name, ncols=2):
    images = [Image.open(OUTDIR / name).convert("RGB") for name in panel_names]
    widths, heights = zip(*(im.size for im in images))
    cell_w = max(widths)
    cell_h = max(heights)
    nrows = math.ceil(len(images) / ncols)
    canvas = Image.new("RGB", (cell_w * ncols, cell_h * nrows), "white")
    for idx, im in enumerate(images):
        r, c = divmod(idx, ncols)
        x = c * cell_w + (cell_w - im.size[0]) // 2
        y = r * cell_h + (cell_h - im.size[1]) // 2
        canvas.paste(im, (x, y))
    out = OUTDIR / out_name
    canvas.save(out)
    print(f"saved {out.relative_to(ROOT)}")
    return out


panel_files = [
    "panel_A_btsp_weight_change.png",
    "panel_B_kernel_fixed_points.png",
    "panel_C_learning_dynamics.png",
    "panel_D_architecture.png",
    "panel_E_receptive_field_formation.png",
    "panel_F_space_sensory_input.png",
    "panel_G_memory_decodability.png",
    "panel_H_decoded_pattern_examples.png",
    "panel_I_decodability_quantification.png",
]
assemble_composite(panel_files, "figure_1_composite.png", ncols=2)
```
